# 🟠 Part A: Matplotlib Questions

### Question 1 — Line Plot  `Matplotlib`

Calculate the **average marks** of all students in each of the four tests (`Test_1` to `Test_4`). Create a **line plot** to show how the class average changes from test to test. Add a title and axis labels.

```python
# ✏️ Your code here
avg = df[['Test_1', 'Test_2', 'Test_3', 'Test_4']].mean(axis=0)
plt.figure(figsize=(5, 4))
plt.plot(avg.index, avg.values, marker='o',color='purple')
plt.title('Average of Class')
plt.xlabel('Test')
plt.ylabel('Average Marks')
plt.show()
```

### Question 2 — Bar Chart  `Matplotlib`

Create a **bar chart** comparing the final `Marks` of the **first five students** in the dataset.

```python
# ✏️ Your code here
df=df.head(5)
plt.bar(df['Student'], df['Marks'])
plt.title('Bar Chart(Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.show()
```

### Question 3 — Histogram  `Matplotlib`

Create a **histogram** of the final `Marks` of all 30 students. Change the **number of bins** (for example 3, 5 and 10) and write what difference you observe.

```python
# ✏️ Your code here
plt.hist(df['Marks'],bins=3,edgecolor='black')
plt.title('Histogram')
plt.xlabel('marks')
plt.show()
plt.hist(df['Marks'],bins=5,edgecolor='black')
plt.title('Histogram')
plt.xlabel('marks')
plt.show()
plt.hist(df['Marks'],bins=10,edgecolor='black')
plt.title('Histogram')
plt.xlabel('marks')
plt.show()
```

**🔎 Observation:** *(it grouped the data,width of the bar decreases)*

### Question 4 — Scatter Plot  `Matplotlib`

Create a **scatter plot** showing the relationship between `Study_Hours` and `Marks`. Does more study time lead to higher marks?

```python
# ✏️ Your code here
plt.scatter(df['Study_Hours'],df['Marks'])
plt.title('Study_Hour vs Marks')
plt.xlabel('Study_Hour')
plt.ylabel('Marks')
plt.show()
```

### Question 5 — Pie Chart  `Matplotlib`

Create a **pie chart** showing the **percentage of students** who prefer each programming language (`Preferred_Language`). Display the percentage on each slice.

```python
# ✏️ Your code here
lang_avg=df['Preferred_Language'].value_counts()
plt.pie(lang_avg,labels=lang_avg.index,autopct='%1.2f%%')
plt.show()
```

# 🟢 Part B: Seaborn Questions

### Question 6 — Count Plot  `Seaborn`

Using **Seaborn**, create a **count plot** showing the number of students in each `Class`.

```python
# ✏️ Your code here
sns.countplot(data=df,x=df['Class'])
plt.title('CountPlot')
plt.show()
```

### Question 7 — Box Plot  `Seaborn`

Using **Seaborn**, create a **box plot** to compare the final `Marks` of **Class A** and **Class B**. Which class has the higher median?

```python
# ✏️ Your code here
#Class A has higher median
sns.boxplot(data=df,x=df['Class'],y=df['Marks'])
plt.title('Class vs Marks')
plt.show()
```

### Question 8 — Heatmap  `Seaborn`

Create a **heatmap** of the correlation between `Study_Hours`, `Attendance`, `Sleep_Hours` and `Marks`. Show the values on the heatmap. Which column has the strongest relationship with `Marks`?

```python
# ✏️ Your code here
numeric_df=df[['Study_Hours', 'Attendance', 'Sleep_Hours', 'Marks']]
correlation=numeric_df.corr()
sns.heatmap(correlation,annot=True,cmap='coolwarm')
plt.show()
```

### Question 9 — Explore the Data  `Matplotlib / Seaborn`



**Plot 1 — Bar plot: average marks by preferred language**

```python
lang_avg=df.groupby('Preferred_Language')['Marks'].mean()
plt.figure(figsize=(4,5))
sns.barplot(x=lang_avg.index,y=lang_avg.values)
plt.ylabel('Avg_Marks')
plt.show()
```

**🔎 Observation 1:** *(It shows the average marks by preferred language)*

**Plot 2 — Scatter plot: sleep hours vs marks**

```python
# ✏️ Your code here
sns.scatterplot(data=df,x=df['Sleep_Hours'],y=df['Marks'])
plt.show()
```

**🔎 Observation 2:** *(Lower sleep_hours gets higher marks)*

**Plot 3 — Line plot: test averages for each class**

```python
# ✏️ Your code here
test_avg=df.groupby('Class')[['Test_1','Test_2','Test_3','Test_4']].mean()
sns.lineplot(data=test_avg,marker='o')
plt.ylabel('Test_mean')
plt.show()
```
